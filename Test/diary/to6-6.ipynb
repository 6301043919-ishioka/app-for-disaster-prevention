# SQLの `JOIN` と `ON` の学習事項まとめ

## 今回の問題

`Sales` テーブルには、売上情報が入っている。

| sale_id | product_id | year | quantity | price |
|---|---|---|---|---|
| 1 | 100 | 2008 | 10 | 5000 |
| 2 | 100 | 2009 | 12 | 5000 |
| 7 | 200 | 2011 | 15 | 9000 |

`Product` テーブルには、商品IDと商品名が入っている。

| product_id | product_name |
|---|---|
| 100 | Nokia |
| 200 | Apple |
| 300 | Samsung |

求める結果は、売上ごとの

| product_name | year | price |
|---|---|---|

である。

---

## 解答SQL

    SELECT
        p.product_name,
        s.year,
        s.price
    FROM Sales s
    JOIN Product p
    ON s.product_id = p.product_id;

---

## `JOIN` の役割

`JOIN` は、複数のテーブルを結合するために使う。

今回の場合、`Sales` テーブルには `product_id` はあるが、`product_name` はない。

例：

| product_id | year | price |
|---|---|---|
| 100 | 2008 | 5000 |

一方、`Product` テーブルには `product_id` に対応する `product_name` がある。

| product_id | product_name |
|---|---|
| 100 | Nokia |

そのため、`JOIN` を使って `Sales` と `Product` を結合する。

    FROM Sales s
    JOIN Product p

これは、

「Sales テーブルに Product テーブルを結合する」

という意味。

---

## `ON` の役割

`ON` は、どの条件でテーブル同士を結合するかを指定する。

今回の条件は、

    ON s.product_id = p.product_id

である。

これは、

「Sales の product_id と Product の product_id が同じ行を結合する」

という意味。

---

## 結合のイメージ

### Sales

| sale_id | product_id | year | price |
|---|---|---|---|
| 1 | 100 | 2008 | 5000 |

### Product

| product_id | product_name |
|---|---|
| 100 | Nokia |

`product_id = 100` が一致しているので、この2つの行が結合される。

### 結果

| product_name | year | price |
|---|---|---|
| Nokia | 2008 | 5000 |

---

## 別名 `s` と `p` について

SQLでは、テーブルに短い別名をつけることができる。

    FROM Sales s
    JOIN Product p

この場合、

| 別名 | 本来のテーブル |
|---|---|
| s | Sales |
| p | Product |

という意味になる。

そのため、

    s.product_id

は、

「Sales テーブルの product_id」

という意味。

    p.product_id

は、

「Product テーブルの product_id」

という意味。

---

## なぜ `s.product_id` や `p.product_id` と書くのか

`product_id` は `Sales` テーブルにも `Product` テーブルにも存在する。

そのため、ただ

    product_id

と書くと、どちらのテーブルの `product_id` なのか分かりにくい。

そこで、

    s.product_id
    p.product_id

のように、テーブル名または別名をつけて区別する。

---

## 今回のSQLの意味

    SELECT
        p.product_name,
        s.year,
        s.price
    FROM Sales s
    JOIN Product p
    ON s.product_id = p.product_id;

このSQLは、

「Sales テーブルと Product テーブルを product_id が同じもの同士で結合し、商品名・年・価格を表示する」

という意味。

---

## 重要ポイント

| SQL | 役割 |
|---|---|
| JOIN | テーブルを結合する |
| ON | 結合する条件を指定する |
| s | Sales テーブルの別名 |
| p | Product テーブルの別名 |
| s.product_id = p.product_id | product_id が同じ行同士を結合する条件 |

---

## まとめ

`JOIN` は、別々のテーブルをくっつけるために使う。

`ON` は、どの列を基準にしてくっつけるかを指定する。

今回なら、

    JOIN Product p
    ON s.product_id = p.product_id

は、

「Sales テーブルに Product テーブルを結合する。ただし、product_id が同じ行だけを結合する」

という意味である。

# SQLで重複を消すコマンド：DISTINCT

## 基本

SQLで重複したデータを消して、同じ値を1回だけ表示したいときは `DISTINCT` を使う。

## 書き方

    SELECT DISTINCT カラム名
    FROM テーブル名;

## 今回の例

    SELECT DISTINCT
        author_id AS id
    FROM Views
    WHERE author_id = viewer_id
    ORDER BY id;

## 意味

`author_id = viewer_id` の行を探すと、自分の記事を見た著者がわかる。

ただし、同じ著者が何回も自分の記事を見ている場合や、同じ行が重複している場合、そのままだと同じ `author_id` が複数回出る可能性がある。

そこで、

    SELECT DISTINCT author_id

とすることで、同じ `author_id` を1回だけ表示できる。

## 例

元のデータ：

| author_id | viewer_id |
|---|---|
| 7 | 7 |
| 7 | 7 |
| 4 | 4 |
| 4 | 4 |

普通に取得すると：

| id |
|---|
| 7 |
| 7 |
| 4 |
| 4 |

`DISTINCT` を使うと：

| id |
|---|
| 4 |
| 7 |

## ポイント

`DISTINCT` は、重複した行を1つにまとめるためのコマンド。

特に今回のように、「条件に合う人のIDを一覧で出したい」「同じ人を何回も表示したくない」という場合に使う。

## 注意点

複数カラムに `DISTINCT` を使うと、そのカラムの組み合わせが重複しているかどうかで判定される。

例：

    SELECT DISTINCT author_id, viewer_id
    FROM Views;

この場合、`author_id` だけではなく、`author_id` と `viewer_id` の組み合わせが同じ行だけが重複として消される。

## まとめ

| 書き方 | 意味 |
|---|---|
| `SELECT DISTINCT user_id` | 同じ `user_id` を1回だけ表示 |
| `SELECT DISTINCT author_id AS id` | `author_id` の重複を消して、列名を `id` にする |
| `COUNT(DISTINCT user_id)` | 重複を除いたユーザー数を数える |

今回の問題では、自分の記事を見た著者を1人1回だけ出したいので、`DISTINCT` が必要。

# Pythonのmap関数：学習事項まとめ

## map関数とは

`map()` は、リストなどの複数のデータに対して、同じ処理を一括で行う関数。

例えば、文字列のリストを全部 `int` に変換したいときなどに使う。

---

## 基本形

    map(関数, データ)

例：

    map(int, ["2019", "02", "10"])

これは、リストの中身それぞれに `int()` を適用するという意味。

つまり、

    int("2019")
    int("02")
    int("10")

をまとめてやっている。

---

## 今回のコード

    year, month, day = map(int, date.split("-"))

例えば、

    date = "2019-02-10"

のとき、

    date.split("-")

によって、

    ["2019", "02", "10"]

になる。

しかし、このままだと全部文字列。

    "2019"
    "02"
    "10"

そこで、

    map(int, date.split("-"))

を使うと、それぞれが整数に変換される。

    2019
    2
    10

そして、

    year, month, day = ...

によって、

    year = 2019
    month = 2
    day = 10

となる。

---

## mapを使わない場合

mapを使わないと、次のように書く必要がある。

    parts = date.split("-")

    year = int(parts[0])
    month = int(parts[1])
    day = int(parts[2])

これは少し長い。

---

## mapを使う場合

    year, month, day = map(int, date.split("-"))

1行で書けるので、短くて見やすい。

---

## 注意点

`map()` の結果は、そのままだとリストではない。

例えば、

    result = map(int, ["1", "2", "3"])

これは見た目上リストではない。

リストとして使いたい場合は、

    result = list(map(int, ["1", "2", "3"]))

と書く。

結果：

    [1, 2, 3]

---

## よく使う例

### 文字列を整数に変換する

    nums = list(map(int, ["1", "2", "3"]))

結果：

    [1, 2, 3]

---

### 入力を整数リストにする

競技プログラミングではよく使う。

    nums = list(map(int, input().split()))

入力：

    1 2 3 4

結果：

    [1, 2, 3, 4]

---

## まとめ

| 書き方 | 意味 |
|---|---|
| `map(int, list)` | リストの各要素を整数に変換 |
| `map(str, list)` | リストの各要素を文字列に変換 |
| `list(map(...))` | mapの結果をリストにする |
| `year, month, day = map(int, date.split("-"))` | 日付文字列を年・月・日に分けて整数化する |

## 今回の重要ポイント

    year, month, day = map(int, date.split("-"))

これは、

    "2019-02-10"

を、

    year = 2019
    month = 2
    day = 10

に変換するための便利な書き方。

`map()` は「複数の要素に同じ処理をまとめて適用する関数」と覚える。

## 9. スライディングウィンドウ

### 何をする考え方か

スライディングウィンドウは、**連続する一部分を見ながら、少しずつ右にずらしていく方法**です。

特に、

```python
長さ k の連続部分配列
```

を扱う問題でよく使います。

---

### 例

```python
nums = [1, 12, -5, -6, 50, 3]
k = 4
```

長さ `4` の部分配列は、

```python
[1, 12, -5, -6]
[12, -5, -6, 50]
[-5, -6, 50, 3]
```

です。

毎回 `sum()` で合計を計算すると遅いので、前の合計を使い回します。

---

### 考え方

最初の合計を作る。

```python
window_sum = sum(nums[:k])
```

この場合、

```python
window_sum = 1 + 12 + (-5) + (-6)
           = 2
```

次に、区間を右に1つずらします。

```python
[1, 12, -5, -6]
 ↓
[12, -5, -6, 50]
```

このとき、

```python
新しく入る値 = 50
抜ける値 = 1
```

なので、

```python
window_sum += 50
window_sum -= 1
```

とすれば、新しい区間の合計になります。

---

### コード

```python
class Solution:
    def findMaxAverage(self, nums: list[int], k: int) -> float:
        window_sum = sum(nums[:k])
        max_sum = window_sum

        for i in range(k, len(nums)):
            window_sum += nums[i]
            window_sum -= nums[i - k]
            max_sum = max(max_sum, window_sum)

        return max_sum / k
```

---

### 重要な部分

```python
for i in range(k, len(nums)):
```

これは、`k` 番目の要素から最後まで見るという意味です。

例えば、

```python
nums = [1, 12, -5, -6, 50, 3]
k = 4
```

なら、最初の4個はすでに計算済みなので、

```python
i = 4
```

から始めます。

---

### この2行が一番大事

```python
window_sum += nums[i]
window_sum -= nums[i - k]
```

意味は、

```python
新しく右から入ってきた値を足す
左から抜けた値を引く
```

です。

これにより、毎回全部足し直さなくて済みます。

---

### 計算量

```python
O(n)
```

です。

各要素をほぼ1回ずつ見るだけなので高速です。

## 10. True / False で存在管理

### 何をする考え方か

HashSet の問題では、

```python
ある数字が存在するかどうか
```

を管理したいです。

そこで、リストを使って、

```python
self.data[key] が True なら存在する
self.data[key] が False なら存在しない
```

と考えます。

---

### コード

```python
class MyHashSet:

    def __init__(self):
        self.data = [False] * 1000001

    def add(self, key: int) -> None:
        self.data[key] = True

    def remove(self, key: int) -> None:
        self.data[key] = False

    def contains(self, key: int) -> bool:
        return self.data[key]
```

---

### `__init__`

```python
def __init__(self):
    self.data = [False] * 1000001
```

最初に、全部 `False` のリストを作ります。

```python
[False, False, False, False, ...]
```

`False` は、

```python
まだ存在しない
```

という意味です。

`1000001` 個用意する理由は、問題の制約で、

```python
0 <= key <= 1000000
```

だからです。

`0` から `1000000` まで使うには、合計で `1000001` 個必要です。

---

### `add`

```python
def add(self, key: int) -> None:
    self.data[key] = True
```

これは、`key` を追加する処理です。

例えば、

```python
myHashSet.add(2)
```

を実行すると、

```python
self.data[2] = True
```

になります。

つまり、

```python
2 は存在する
```

という意味です。

---

### `contains`

```python
def contains(self, key: int) -> bool:
    return self.data[key]
```

これは、`key` が存在するか確認する処理です。

例えば、

```python
myHashSet.contains(2)
```

は、

```python
return self.data[2]
```

と同じです。

もし `2` を追加していれば、

```python
True
```

を返します。

追加していなければ、

```python
False
```

を返します。

---

### `remove`

```python
def remove(self, key: int) -> None:
    self.data[key] = False
```

これは、`key` を削除する処理です。

例えば、

```python
myHashSet.remove(2)
```

を実行すると、

```python
self.data[2] = False
```

になります。

つまり、

```python
2 は存在しない
```

という状態に戻します。

---

### 具体例

```python
myHashSet = MyHashSet()

myHashSet.add(1)
myHashSet.add(2)
```

この時点では、

```python
self.data[1] = True
self.data[2] = True
self.data[3] = False
```

というイメージです。

なので、

```python
myHashSet.contains(1)
```

は、

```python
True
```

になります。

一方で、

```python
myHashSet.contains(3)
```

は、

```python
False
```

になります。

次に、

```python
myHashSet.remove(2)
```

をすると、

```python
self.data[2] = False
```

になります。

そのため、

```python
myHashSet.contains(2)
```

は、

```python
False
```

になります。

---

### 計算量

| 操作 | 計算量 |
|---|---:|
| add | O(1) |
| remove | O(1) |
| contains | O(1) |

リストのインデックスに直接アクセスするだけなので、すべて高速です。

# Pythonで「インデックス」と「値」を扱う書き方まとめ

## 1. `enumerate(nums)`

`enumerate(nums)` を使うと、配列の

- インデックス
- 値

を同時に取り出せます。

```python
nums = [3, 6, 1, 0]

for i, num in enumerate(nums):
    print(i, num)
```

出力：

```text
0 3
1 6
2 1
3 0
```

つまり、

```python
for i, num in enumerate(nums):
```

では、

```text
i   → インデックス
num → その位置の値
```

を表します。

---

## 2. `for num in nums`

値だけ使いたいときは、普通の `for` 文で十分です。

```python
nums = [3, 6, 1, 0]

for num in nums:
    print(num)
```

出力：

```text
3
6
1
0
```

この書き方では、インデックスは取り出せません。

---

## 3. `range(len(nums))`

インデックスを使って配列にアクセスしたいときに使います。

```python
nums = [3, 6, 1, 0]

for i in range(len(nums)):
    print(i, nums[i])
```

出力：

```text
0 3
1 6
2 1
3 0
```

この場合、

```python
i
```

がインデックスで、

```python
nums[i]
```

がその位置の値です。

---

## 4. `enumerate(nums)` と `range(len(nums))` の違い

次の2つは似ています。

### `enumerate(nums)` の場合

```python
for i, num in enumerate(nums):
    print(i, num)
```

### `range(len(nums))` の場合

```python
for i in range(len(nums)):
    print(i, nums[i])
```

どちらも、インデックスと値を扱えます。

ただし、Pythonでは基本的に

```python
enumerate(nums)
```

の方が読みやすく、よく使われます。

---

## 5. 比較表

| 書き方 | 取れるもの | 使う場面 |
|---|---|---|
| `for num in nums` | 値だけ | インデックスが不要なとき |
| `for i in range(len(nums))` | インデックス | インデックス操作が必要なとき |
| `for i, num in enumerate(nums)` | インデックスと値 | 両方ほしいとき |

ここでは、

です。

```python
nums = [3, 6, 1, 0]
```

なら、ループでは次のように取り出されます。

| ループ回数 | `i` | `num` |
|---:|---:|---:|

```python
if i != max_index
```

は、

```text
今見ている場所が、最大値の場所ではないなら
```

```python
max_num >= num * 2
```

です。

逆に、

```python
max_num < num * 2
```

なら条件を満たさないので、

```python
return -1
```

します。

# `ord(char) - ord("a")` の詳しい解説

## 1. 何をしている処理か

```python
ord(char) - ord("a")
```

これは、英小文字を **アルファベット順の番号** に変換する処理です。

具体的には、次のように変換します。

| 文字 | 番号 |
|---|---:|
| `"a"` | 0 |
| `"b"` | 1 |
| `"c"` | 2 |
| `"d"` | 3 |
| ... | ... |
| `"z"` | 25 |

つまり、`widths` の配列で対応する文字幅を取り出すために使います。

---

# 2. なぜ番号に変換する必要があるのか

今回の問題では、`widths` は次のような配列です。

```python
widths = [4, 10, 10, 10, ...]
```

この配列は、アルファベットと次のように対応しています。

| 配列の位置 | 文字 | 意味 |
|---:|---|---|
| `widths[0]` | `"a"` | `"a"` の幅 |
| `widths[1]` | `"b"` | `"b"` の幅 |
| `widths[2]` | `"c"` | `"c"` の幅 |
| `widths[3]` | `"d"` | `"d"` の幅 |
| ... | ... | ... |
| `widths[25]` | `"z"` | `"z"` の幅 |

だから、たとえば文字が `"c"` なら、

```python
widths[2]
```

を取り出したいです。

そのために、`"c"` を `2` に変換する必要があります。

そこで使うのが、

```python
ord(char) - ord("a")
```

です。

---

# 3. `ord()` とは何か

`ord()` は、文字を数値に変換する関数です。

```python
ord("a")
```

と書くと、`"a"` に対応するUnicode番号を返します。

英小文字の場合、次のようになっています。

# 4. なぜ `ord("a")` を引くのか

```python
ord("a") = 97
ord("b") = 98
ord("c") = 99
```

```python
widths[0]  # a
widths[1]  # b
widths[2]  # c
```

になります。

になってしまいます。

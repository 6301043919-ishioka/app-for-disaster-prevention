# ビット演算子まとめ（Python）

## 1. ビットとは

ビットとは、コンピュータが扱う最小単位で、`0` または `1` のこと。

例えば、10進数の `5` は2進数ではこう表せる。

```text

```

10進数の `6` は、

```text
6 = 110
```

ビット演算は、この `0` と `1` を直接操作する演算。

# 9. Reverse Bitsで使った処理

LeetCodeの Reverse Bits では、この2行が重要。

```python
result = (result << 1) | (n & 1)
n = n >> 1
```

## 意味

```python
n & 1
```

で、`n` の右端のビットを取り出す。

```python
result << 1
```

で、`result` を左に1つずらして、新しいビットを入れる場所を作る。

```python
| (n & 1)
```

で、取り出したビットを `result` の右端に追加する。

```python
n = n >> 1
```

で、`n` を右にずらして次のビットを見る。

---

# 10. Reverse Bitsの完成コード

```python
class Solution:
    def reverseBits(self, n: int) -> int:
        result = 0

        for _ in range(32):
            result = (result << 1) | (n & 1)
            n = n >> 1

        return result
```

## 処理の流れ

```text
32回くり返す
↓
nの右端ビットを取り出す
↓
resultを左にずらす
↓
取り出したビットをresultの右端に追加する
↓
nを右にずらす
```

## `sort()` と `sorted()` の違いまとめ

`sorted()` は、並び替えた結果を **新しいリストとして返す関数**です。  
元のデータは変更されません。

```python
nums = [3, 1, 2]

result = sorted(nums)

print(result)  # [1, 2, 3]
print(nums)    # [3, 1, 2]
```

文字列にも使えます。

```python
s = "anagram"

print(sorted(s))
# ['a', 'a', 'a', 'g', 'm', 'n', 'r']
```

`.sort()` は、リスト自体を直接並び替えるメソッドです。  
元のリストが変更されます。

```python
nums = [3, 1, 2]

nums.sort()

print(nums)  # [1, 2, 3]
```

## 違いのまとめ

| 項目 | `sorted()` | `.sort()` |
|---|---|---|
| 種類 | 関数 | リストのメソッド |
| 元のリスト | 変更しない | 変更する |
| 戻り値 | 並び替えた新しいリスト | `None` |
| 文字列に使えるか | 使える | 使えない |

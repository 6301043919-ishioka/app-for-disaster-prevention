# Pythonの `set()` 関数まとめ

## 1. `set()` とは？

`set()` は、Pythonで **集合** を作る関数です。

集合とは、数学の集合と同じように、

- 重複がない
- 順番を基本的に気にしない
- 値が入っているかを高速に調べられる

という特徴があります。

---

## 2. 基本形

```python
set(リストや文字列など)
```

例えば、リストを `set()` に入れると、重複を消した集合を作れます。

```python
nums = [1, 2, 2, 3, 3, 3]

nums_set = set(nums)

print(nums_set)
```

出力：

```python
{1, 2, 3}
```

`2` や `3` が何回出てきても、`set` では1回だけになります。

---

## 3. `set()` の一番大事な特徴

### 重複を消せる

```python
nums = [4, 3, 2, 7, 8, 2, 3, 1]

nums_set = set(nums)

print(nums_set)
```

出力例：

```python
{1, 2, 3, 4, 7, 8}
```

元のリスト：

```python
[4, 3, 2, 7, 8, 2, 3, 1]
```

には `2` と `3` が2回出ています。

でも、`set()` にすると重複が消えて、

```python
{1, 2, 3, 4, 7, 8}
```

になります。

---

## 4. `set` は存在チェックに強い

`set` は、

```python
値 in set
```

で、その値が入っているかを高速に調べられます。

```python
nums_set = {1, 2, 3, 4}

print(3 in nums_set)
print(5 in nums_set)
```

出力：

```python
True
False
```

意味：

```python
3 in nums_set
```

は、「`3` は `nums_set` の中にある？」という意味です。

```python
5 in nums_set
```

は、「`5` は `nums_set` の中にある？」という意味です。

---

## 5. `not in` もよく使う

```python
nums_set = {1, 2, 3, 4}

print(5 not in nums_set)
```

出力：

```python
True
```

これは、

```python
5はnums_setの中にない
```

という意味です。

---

## 6. 今回の問題での使い方

問題：

```python
nums = [4, 3, 2, 7, 8, 2, 3, 1]
```

この中で、`1` から `8` までの数字のうち、出てこない数字を探したい。

まず、`set()` を使って、出てきた数字を集合にします。

```python
nums_set = set(nums)
```

すると、

```python
{1, 2, 3, 4, 7, 8}
```

になります。

つまり、

```python
5
6
```

が入っていないことがわかります。

---

## 7. コード例

```python
class Solution:
    def findDisappearedNumbers(self, nums: List[int]) -> List[int]:
        nums_set = set(nums)
        answer = []

        for i in range(1, len(nums) + 1):
            if i not in nums_set:
                answer.append(i)

        return answer
```

---

## 8. コードの流れ

例えば、

```python
nums = [4, 3, 2, 7, 8, 2, 3, 1]
```

まず、

```python
nums_set = set(nums)
```

で、

```python
{1, 2, 3, 4, 7, 8}
```

を作る。

次に、

```python
for i in range(1, len(nums) + 1):
```

で、`1` から `8` まで確認する。

```python
1 → ある
2 → ある
3 → ある
4 → ある
5 → ない
6 → ない
7 → ある
8 → ある
```

ない数字だけを `answer` に追加するので、

```python
[5, 6]
```

が答えになる。

---

## 9. `list` と `set` の違い

| 種類 | 特徴 |
|---|---|
| `list` | 順番を保つ、重複OK |
| `set` | 重複なし、存在チェックが速い、順番は基本気にしない |

---

## 10. 注意点

`set` は順番を保証しません。

```python
nums = [3, 1, 2]
print(set(nums))
```

出力は、

```python
{1, 2, 3}
```

のようになることがあります。

元の順番：

```python
[3, 1, 2]
```

とは違う順番で表示されることがあります。

なので、

- 順番が大事 → `list`
- 重複を消したい → `set`
- ある値が含まれているか調べたい → `set`

と覚えるとよいです。

---

## 11. まとめ

`set()` は、リストなどから集合を作る関数です。

特に大事なのはこの2つです。

```python
重複を消せる
```

```python
値が入っているかを高速に調べられる
```

今回のように、

```python
1からnまでの中で、numsに出てこない数字を探す
```

問題では、`set()` を使うとかなり簡単に解けます。

# Pythonの `upper()` 関数まとめ

結果：

```python
"ABC"
```

## 2. 基本形

```python
文字列.upper()
```

例：

```python
s = "hello"
print(s.upper())
```

出力：

```python
HELLO
```

---

## 3. 小文字だけが大文字になる

```python
s = "abcDEF"
print(s.upper())
```

出力：

```python
ABCDEF
```

すでに大文字の文字は、そのままです。

---

## 4. 数字や記号は変わらない

```python
s = "a1-b2-c3"
print(s.upper())
```

出力：

数字や `-` などの記号は、大文字・小文字の概念がないので変化しません。

文字列全体だけでなく、1文字に対しても使えます。

出力：

---

出力：

```python
10
1
```

# 5. 今回のコードでの意味

今回のコードでは、こう使っていました。

```python
max_count = max(max_count, count)
```

これは、

```python
max_count と count を比べて、大きい方を max_count に入れる
```

という意味です。

---

## 例

```python
max_count = 2
count = 3

max_count = max(max_count, count)

print(max_count)
```

出力：

```python
3
```

`2` と `3` を比べて、大きい方の `3` が `max_count` に入ります。

# 6. `if` 文で書くとこうなる

```python
max_count = max(max_count, count)
```

これは、次のコードと同じ意味です。

```python
if count > max_count:
    max_count = count
```

つまり、

```python
今の連続数 count が、過去最大 max_count より大きければ更新する
```

ということです。

---

# 7. `min()` も同じ考え方

```python
smallest = min(smallest, num)
```

これは、

```python
smallest と num を比べて、小さい方を smallest に入れる
```

という意味です。

---

## 例

```python
smallest = 10
num = 4

smallest = min(smallest, num)

print(smallest)
```

出力：

```python
4
```

---

# 8. 文字列にも使える

```python
print(max("abc"))
print(min("abc"))
```

出力：

```python
c
a
```

文字列の場合、アルファベット順のような基準で比べます。

```python
"a" < "b" < "c"
```

なので、

```python
max("abc")  # "c"
min("abc")  # "a"
```

になります。

---

# 9. よく使う場面

## 最大値を更新したいとき

```python
max_count = max(max_count, count)
```

例：連続する1の最大数、最大得点、最大利益など。

---

## 最小値を更新したいとき

```python
min_value = min(min_value, num)
```

例：最小値、最短距離、最小コストなど。

---

# 10. 注意点

空のリストに対して `max()` や `min()` を使うとエラーになります。

これはエラーになります。

```python
空のリストには最大値も最小値も存在しないから
```

です。

# `set()` 関数を使うタイミングまとめ

`set()` は、リストや文字列などを **集合** に変換する関数です。

集合にすると、主に次のような特徴があります。

- 重複が消える
- 要素の順番は基本的に保持されない
- 「含まれているか」の確認が速い
- 共通部分・差集合・部分集合の判定ができる

---

# 1. 重複を消したいとき

同じ値が何回も出てくるリストから、重複をなくしたいときに使います。

```python
nums = [1, 2, 2, 3, 3, 3]

unique_nums = set(nums)

print(unique_nums)
```

出力は、

```python
{1, 2, 3}
```

になります。

`2` や `3` が複数回出ていても、`set` にすると1つだけになります。

---

# 2. 文字の種類を調べたいとき

文字列にどんな文字が使われているかを調べたいときにも使えます。

```python
word = "Alaska"

letters = set(word.lower())

print(letters)
```

出力は、

```python
{'a', 'l', 's', 'k'}
```

になります。

`"Alaska"` は本当は6文字ですが、`a` が重複しているので、`set` にすると文字の種類だけが残ります。

---

# 3. 「含まれているか」を調べたいとき

ある要素が存在するかを調べたいときにも `set` は便利です。

```python
row1 = set("qwertyuiop")

print("q" in row1)
print("a" in row1)
```

出力は、

```python
True
False
```

になります。

`q` は1行目にあるので `True`、`a` は1行目にないので `False` です。

---

# 4. 共通部分を調べたいとき

2つの配列に共通している要素を調べたいときは、`&` を使います。

```python
nums1 = [1, 2, 2, 3]
nums2 = [2, 3, 4]

common = set(nums1) & set(nums2)

print(common)
```

出力は、

```python
{2, 3}
```

になります。

`nums1` と `nums2` の両方にある数字は `2` と `3` です。

---

# 5. 差を調べたいとき

片方にはあるけど、もう片方にはない要素を調べたいときは `-` を使います。

```python
all_nums = set([1, 2, 3, 4, 5])
existing = set([1, 2, 4])

missing = all_nums - existing

print(missing)
```

出力は、

```python
{3, 5}
```

になります。

`all_nums` にはあるけど、`existing` にはない数字が取り出されます。

---

# 6. 部分集合かどうかを調べたいとき

ある集合の要素が、すべて別の集合に含まれているかを調べるときは `<=` を使います。

```python
letters = set("dad")
row2 = set("asdfghjkl")

print(letters <= row2)
```

出力は、

```python
True
```

になります。

`dad` に使われている文字は、

```python
{'d', 'a'}
```

です。

`d` も `a` もキーボードの2行目にあるので、`True` になります。

---

# 今回のキーボード問題での使い方

```python
class Solution:
    def findWords(self, words):
        row1 = set("qwertyuiop")
        row2 = set("asdfghjkl")
        row3 = set("zxcvbnm")

        answer = []

        for word in words:
            letters = set(word.lower())

            if letters <= row1 or letters <= row2 or letters <= row3:
                answer.append(word)

        return answer
```

## 大文字・小文字判定で使う3つの関数

### `isupper()`
文字列が **すべて大文字** なら `True`

例：
```python
"USA".isupper()   # True
"Usa".isupper()   # False
```

### `islower()`
文字列が **すべて小文字** なら `True`

例：
```python
"leetcode".islower()   # True
"Leetcode".islower()   # False
```

### `istitle()`
文字列が **先頭だけ大文字で、残りが小文字** なら `True`

例：
```python
"Google".istitle()   # True
"FlaG".istitle()     # False
```

## 今回の問題での使い方

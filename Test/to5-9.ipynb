# `for factor in [2, 3, 5]:` の解説

```python
for factor in [2, 3, 5]:
```

これは、Pythonで **リストの中身を1つずつ取り出して、同じ処理を繰り返す** ための書き方。

---

## 1. `[2, 3, 5]` はリスト

```python
[2, 3, 5]
```

これは、複数の値をまとめた **リスト**。

この中には、

```text
2
3
5
```

の3つの値が入っている。

---

## 2. `for` は繰り返し処理

```python
for factor in [2, 3, 5]:
```

これは日本語にすると、

```text
[2, 3, 5] の中から1つずつ取り出して、
それを factor に入れて、
下の処理を繰り返す
```

という意味。

---

## 3. `factor` の中身は順番に変わる

このコードでは、`factor` は次のように変化する。

```text
1回目: factor = 2
2回目: factor = 3
3回目: factor = 5
```

つまり、Pythonの内部では次のようなイメージ。

```python
factor = 2
# 下の処理を実行

factor = 3
# 下の処理を実行

factor = 5
# 下の処理を実行
```

`factor` に `2` が入り、次に `3` が入り、最後に `5` が入る。

という処理。

この時点で、

```python
n = 15
```

```python
15 % 3 == 0
```

なので、3で割る。

```python
15 // 3 = 5
```

この時点で、

```python
n = 5
```

---

### 3回目：factor = 5

```python
5 % 5 == 0
```

なので、5で割る。

```python
5 // 5 = 1
```

この時点で、

```python
n = 1
```

最後に `n == 1` なので、`30` は ugly number と判断できる。

と書いた時点で、Pythonが自動的に `factor` という変数を作る。

# Pythonの `zip()` 関数まとめ

## `zip()` とは

`zip()` は、Pythonにもともと用意されている関数です。

複数のリスト・文字列・タプルなどを、**同じ位置同士でペアにして取り出す**ことができます。

---

## 基本形

```python
zip(データ1, データ2)
```

例えば、

```python
pattern = "abba"
words = ["dog", "cat", "cat", "dog"]

for char, word in zip(pattern, words):
    print(char, word)
```

出力は、

```python
a dog
b cat
b cat
a dog
```

になります。

---

## 何が起きているか

```python
pattern = "abba"
words = ["dog", "cat", "cat", "dog"]
```

この2つを `zip()` で組み合わせると、

```python
a と dog
b と cat
b と cat
a と dog
```

のように、同じ順番のもの同士がペアになります。

つまり、

| patternの文字 | wordsの単語 |
|---|---|
| a | dog |
| b | cat |
| b | cat |
| a | dog |

という対応になります。

---

## 今回の問題での使い方

```python
for char, word in zip(pattern, words):
```

これは、

```python
patternの1文字
```

と

```python
sをsplitした単語
```

を1組ずつ取り出すという意味です。

例えば、

```python
pattern = "abba"
s = "dog cat cat dog"
words = s.split()
```

このとき、

```python
words = ["dog", "cat", "cat", "dog"]
```

なので、

```python
for char, word in zip(pattern, words):
    print(char, word)
```

は、

```python
a dog
b cat
b cat
a dog
```

となります。

---

## `char` と `word` は自分で決めた変数名

```python
for char, word in zip(pattern, words):
```

の `char` と `word` は、Pythonで決まっている名前ではありません。

自分で決めた変数名です。

例えば、次のように書いても同じ意味です。

```python
for x, y in zip(pattern, words):
    print(x, y)
```

ただし、意味がわかりやすいので、

```python
char
word
```

のように書いています。

---

## `zip()` は短い方に合わせる

`zip()` は、2つのデータの長さが違う場合、**短い方に合わせて止まります**。

例えば、

```python
letters = "abc"
words = ["dog", "cat"]

for letter, word in zip(letters, words):
    print(letter, word)
```

出力は、

```python
a dog
b cat
```

になります。

`c` は対応する単語がないので無視されます。

---

## だから先に長さを確認する

今回の問題では、

```python
pattern = "abba"
s = "dog cat cat"
```

のように、文字数と単語数が違う場合があります。

この場合、

```python
patternの長さ = 4
wordsの長さ = 3
```

なので、そもそも正しく対応できません。

だから、先に次のように確認します。

```python
if len(pattern) != len(words):
    return False
```

これを入れておくことで、`zip()` が短い方に合わせて途中で止まってしまう問題を防げます。

---

## リストに変換して中身を見ることもできる

`zip()` の結果を見たいときは、`list()` を使います。

```python
pattern = "abba"
words = ["dog", "cat", "cat", "dog"]

print(list(zip(pattern, words)))
```

出力は、

```python
[('a', 'dog'), ('b', 'cat'), ('b', 'cat'), ('a', 'dog')]
```

になります。

---

## まとめ

`zip()` は、複数のデータを同じ位置同士でペアにする関数。

```python
zip(pattern, words)
```

と書くと、

```python
patternの1文字
wordsの1単語
```

をセットで取り出せる。

今回の問題では、

```python
for char, word in zip(pattern, words):
```

によって、

```python
a dog
b cat
b cat
a dog
```

のように順番に対応を確認している。

ただし、`zip()` は短い方に合わせて止まるので、今回のような問題では先に

```python
if len(pattern) != len(words):
    return False
```

を書くのが重要。

# Pythonの `get()` 関数まとめ

## `get()` とは？

`get()` は、Pythonの辞書 `dict` から値を取り出すためのメソッドです。

普通に辞書から値を取り出す場合は、次のように書きます。

```python
count[char]
```

ただし、この書き方だと、`char` が辞書の中に存在しない場合にエラーになります。

そこで使うのが `get()` です。

```python
count.get(char, 0)
```

これは、

```text
char が辞書にある場合 → その値を返す
char が辞書にない場合 → 0 を返す
```

という意味です。

---

## 基本形

```python
辞書.get(キー, キーが存在しないときの値)
```

例：

```python
count = {"a": 2, "b": 1}

print(count.get("a", 0))  # 2
print(count.get("c", 0))  # 0
```

### 解説

```python
count.get("a", 0)
```

`"a"` は辞書に存在するので、値の `2` が返ります。

```python
count.get("c", 0)
```

`"c"` は辞書に存在しないので、代わりに `0` が返ります。

---

## 今回のコードでの使い方

```python
count[char] = count.get(char, 0) + 1
```

これは、文字の出現回数を数えるための処理です。

意味は次の通りです。

```text
char がすでに count にある場合
→ 今の回数に +1 する

char がまだ count にない場合
→ 0 から始めて +1 する
```

---

## 具体例

例えば、

```python
s = "leetcode"
count = {}
```

最初の文字 `"l"` を見るとします。

```python
count.get("l", 0)
```

まだ `"l"` は辞書にないので、`0` が返ります。

そのため、

```python
count["l"] = 0 + 1
```

となり、

```python
count = {"l": 1}
```

になります。

次に `"e"` を見ると、まだ `"e"` も辞書にないので、

```python
count = {"l": 1, "e": 1}
```

になります。

その後、もう一度 `"e"` が出てきた場合、

```python
count.get("e", 0)
```

は、すでに `"e": 1` があるので `1` を返します。

だから、

```python
count["e"] = 1 + 1
```

となり、

```python
count = {"l": 1, "e": 2}
```

になります。

---

## `get()` を使わない場合

`get()` を使わないと、次のように書く必要があります。

```python
count = {}

for char in s:
    if char in count:
        count[char] += 1
    else:
        count[char] = 1
```

これは、

```text
すでにある文字なら +1
まだない文字なら 1 にする
```

という処理です。

---

## `get()` を使うと短く書ける

上のコードは、`get()` を使うと次のように短くできます。

```python
count = {}

for char in s:
    count[char] = count.get(char, 0) + 1
```

---

## まとめ

```python
count.get(char, 0)
```

は、

```text
char が辞書にあるなら、その値を返す
char が辞書にないなら、0 を返す
```

という意味です。

今回の問題では、文字が何回出てきたかを数えるために使っています。

```python
count[char] = count.get(char, 0) + 1
```

この1行は、

```text
今までの出現回数を取り出して、それに1を足して保存する
```

という処理です。

## 進数変換の王道パターン①：割って余りを見る方法

ある整数を、別の進数の文字列に変換するときの基本は、

```python
num % base
num //= base
```

を使う方法です。

---

## 考え方

例えば、10進数の `26` を16進数に変換する場合を考えます。

16進数では、1桁で `0〜15` を表します。

```text
10進数   16進数
0        0
1        1
2        2
...
9        9
10       a
11       b
12       c
13       d
14       e
15       f
```

---

## 変換の流れ

```text
26 ÷ 16 = 1 あまり 10
1 ÷ 16  = 0 あまり 1
```

余りを下から読むので、

```text
1a
```

になります。

つまり、

```text
26 = 1a
```

です。

---

## なぜ余りを見るのか

```python
num % base
```

は、その進数で見たときの **一番右の桁** を取り出す処理です。

例えば、

```python
26 % 16
```

は、

```text
10
```

なので、16進数の一番右の桁は `a` になります。

次に、

```python
num //= base
```

で、すでに調べた一番右の桁を捨てます。

```python
26 // 16
```

は、

```text
1
```

なので、次は `1` を調べます。

---

## 基本形

```python
while num > 0:
    digit = num % base
    result = chars[digit] + result
    num //= base
```

意味はこうです。

```text
num % base
→ 一番右の桁を取り出す

chars[digit]
→ その数字に対応する文字を取り出す

result = chars[digit] + result
→ 右から求めた桁を、答えの左側に追加する

num //= base
→ 次の桁を見るために、baseで割る
```

---

## 16進数変換のコード

```python
class Solution:
    def toHex(self, num: int) -> str:
        if num == 0:
            return "0"

        chars = "0123456789abcdef"

        # 負の数を32ビットの2の補数表現として扱う
        num = num & 0xffffffff

        result = ""

        while num > 0:
            digit = num % 16
            result = chars[digit] + result
            num //= 16

        return result
```

---

## `num = 26` の具体例

```text
num = 26
result = ""

1回目:
26 % 16 = 10
chars[10] = "a"
result = "a"
num = 26 // 16 = 1

2回目:
1 % 16 = 1
chars[1] = "1"
result = "1" + "a" = "1a"
num = 1 // 16 = 0

終了
```

答えは、

```text
"1a"
```

です。

---

## この方法のポイント

この方法は、16進数だけでなく、いろいろな進数に使えます。

```text
2進数にしたい  → base = 2
8進数にしたい  → base = 8
16進数にしたい → base = 16
```

つまり、

```python
num % base
num //= base
```

を使えば、基本的に任意の進数へ変換できます。

---

## まとめ

```text
ある整数をN進数に変換する王道パターンは、

1. num % N で一番右の桁を取り出す
2. num //= N で次の桁へ進む
3. 取り出した桁は右から順に出てくる
4. だから答えには左側に追加する
```

16進数なら、

```python
digit = num % 16
num //= 16
```

を使います。

# `ord()` 関数まとめ

## 1. `ord()` とは

`ord()` は、**1文字を文字コードの数字に変換する関数**。

基本形はこれ。

```python
ord(文字)
```

例えば、

```python
print(ord('A'))
print(ord('a'))
print(ord('0'))
```

出力は、

```python
65
97
48
```

になる。

つまり、`ord()` は文字そのものを数字にするのではなく、  
その文字に割り当てられている **文字コード** を返す。

---

## 2. 数字の文字も文字コードに変換される

例えば、

```python
print(ord('0'))
print(ord('1'))
print(ord('2'))
print(ord('3'))
```

出力は、

```python
48
49
50
51
```

になる。

ここで大事なのは、`'0'`, `'1'`, `'2'` は **文字列** だということ。

```python
'5'
```

は数字の `5` ではなく、文字としての `"5"`。

だから、

```python
ord('5')
```

をすると、

```python
53
```

になる。

---

## 3. なぜ `ord('5') - ord('0')` で数字の `5` になるのか

数字文字の文字コードは連続している。

```python
ord('0')  # 48
ord('1')  # 49
ord('2')  # 50
ord('3')  # 51
ord('4')  # 52
ord('5')  # 53
```

だから、

```python
ord('5') - ord('0')
```

は、

```python
53 - 48
```

なので、

```python
5
```

になる。

つまり、

```python
ord('5') - ord('0')
```

は、

```text
文字としての '5' を、数字の 5 に変換している
```

という意味。

---

## 4. 具体例

```python
c = '7'

num = ord(c) - ord('0')

print(num)
```

出力は、

```python
7
```

になる。

流れはこう。

```text
ord('7') は 55
ord('0') は 48

55 - 48 = 7
```

だから、文字の `'7'` を数字の `7` として扱える。

---

## 5. この問題で使う理由

今回の問題は、文字列で表された数字を足す問題。

例えば、

```python
num1 = "456"
num2 = "77"
```

このとき、`num1[2]` は、

```python
'6'
```

になる。

でもこれは文字なので、そのまま足し算すると数字の足し算にはならない。

そこで、

```python
digit1 = ord(num1[i]) - ord('0')
```

とする。

例えば、

```python
num1 = "456"
i = 2

digit1 = ord(num1[i]) - ord('0')
```

これは、

```python
digit1 = ord('6') - ord('0')
```

なので、

```python
digit1 = 54 - 48
digit1 = 6
```

になる。

つまり、

```text
文字の '6' を、数字の 6 に変換している
```

ということ。

---

## 6. `int()` との違い

実は、1文字だけなら次のようにも書ける。

```python
digit = int('6')
```

これは普通に、

```python
6
```

になる。

なので、今回の問題では、

```python
digit1 = int(num1[i])
```

と書いてもよい。

ただし、問題文で

```text
入力を整数に直接変換してはいけない
```

とある場合、

```python
int(num1)
```

のように文字列全体を変換するのはダメ。

```python
int("456")
```

はダメ。

一方で、

```python
int("6")
```

のように1桁ずつ変換するのは、筆算の処理として使われることが多い。

---

## 7. `ord()` を使ったコード例

```python
num1 = "456"
num2 = "77"

i = len(num1) - 1
j = len(num2) - 1

carry = 0
ans = []

while i >= 0 or j >= 0 or carry:
    digit1 = ord(num1[i]) - ord('0') if i >= 0 else 0
    digit2 = ord(num2[j]) - ord('0') if j >= 0 else 0

    total = digit1 + digit2 + carry

    ans.append(str(total % 10))
    carry = total // 10

    i -= 1
    j -= 1

print(''.join(ans[::-1]))
```

出力は、

```python
533
```

## 8. `ord()` と反対の関数 `chr()`

`ord()` の反対は `chr()`。

`chr()` は、数字の文字コードから文字を作る関数。

```python
print(chr(65))
print(chr(97))
print(chr(48))
```

出力は、

```python
A
a
0
```

つまり、

```python
ord('A')  # 65
chr(65)   # 'A'
```

という関係。

## 9. まとめ

| 書き方 | 意味 |
|---|---|
| `ord('A')` | `'A'` の文字コードを返す |
| `ord('0')` | `'0'` の文字コードを返す |
| `ord('5') - ord('0')` | 文字 `'5'` を数字 `5` にする |
| `chr(65)` | 文字コード `65` から文字 `'A'` を作る |
| `int('5')` | 文字 `'5'` を数字 `5` にする |

今回の問題では、

```python
ord(num1[i]) - ord('0')
```

が重要。

意味は、

```text
num1[i] に入っている1文字の数字を、本当の数字に変換する
```

ということ。

例えば、

```python
num1[i] = '6'
```

なら、

```python
ord('6') - ord('0')
```

で、

```python
6
```

になる。

## 8. なぜ「10ドル札 + 5ドル札」を優先するのか

例えば、手元に次のようにあるとします。

```python
five = 3
ten = 1
```

20ドル札のお客さんに対して、15ドルのお釣りを返す必要があります。

このとき、もし5ドル札3枚で返すと、

```python
five = 0
ten = 1
```

になります。

すると、次に10ドル札のお客さんが来たとき、  
5ドルのお釣りを返せなくなります。

一方で、10ドル札 + 5ドル札で返すと、

```python
five = 2
ten = 0
```

となり、5ドル札を残せます。

したがって、20ドル札に対しては、

```python
10ドル札 + 5ドル札を優先する
```

のがよいです。

## 9. 完成コード

```python
class Solution:
    def lemonadeChange(self, bills: List[int]) -> bool:
        five = 0
        ten = 0

        for bill in bills:
            # 5ドル札の場合：お釣り不要
            if bill == 5:
                five += 1

            # 10ドル札の場合：5ドルのお釣りが必要
            elif bill == 10:
                if five > 0:
                    five -= 1
                    ten += 1
                else:
                    return False

            # 20ドル札の場合：15ドルのお釣りが必要
            else:
                # 10ドル札 + 5ドル札を優先
                if ten > 0 and five > 0:
                    ten -= 1
                    five -= 1

                # 5ドル札3枚で返す
                elif five >= 3:
                    five -= 3

                # お釣りを作れない
                else:
                    return False

        return True
```

# 学習事項：行列の転置コード

## 1. 転置とは何か

転置とは、行列の **行と列を入れ替える** ことです。

元の行列がこちらだとします。

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]
```

これは **2行3列** の行列です。

転置すると、次のようになります。

```python
[
    [1, 4],
    [2, 5],
    [3, 6]
]
```

元の行列の「列」が、転置後の「行」になります。

```python
元の0列目: 1, 4 → 転置後の0行目: [1, 4]
元の1列目: 2, 5 → 転置後の1行目: [2, 5]
元の2列目: 3, 6 → 転置後の2行目: [3, 6]
```

---

## 2. 転置の本質

転置では、元の行列の要素

```python
matrix[i][j]
```

が、転置後では

```python
result[j][i]
```

に移動します。

つまり、

```python
matrix[i][j] → result[j][i]
```

です。

ここが一番重要です。

---

## 3. 完成コード

```python
from typing import List

class Solution:
    def transpose(self, matrix: List[List[int]]) -> List[List[int]]:
        rows = len(matrix)
        cols = len(matrix[0])

        result = []

        for j in range(cols):
            new_row = []

            for i in range(rows):
                new_row.append(matrix[i][j])

            result.append(new_row)

        return result
```

## 6. reduce を使わずに書くと

```python
from collections import Counter
from math import gcd
from typing import List

class Solution:
    def hasGroupsSizeX(self, deck: List[int]) -> bool:
        counts = Counter(deck)

        g = 0
        for count in counts.values():
            g = gcd(g, count)

        return g > 1
```

```python
g = 0
for count in counts.values():
    g = gcd(g, count)
```

## 7. 最終コード

```python
from typing import List
from collections import Counter
from math import gcd
from functools import reduce

class Solution:
    def hasGroupsSizeX(self, deck: List[int]) -> bool:
        counts = Counter(deck)

        g = reduce(gcd, counts.values())

        return g > 1
```

# Python学習事項：`current = (current << 1) | node.val`

## 結論

```python
current = (current << 1) | node.val
```

これは、**今まで作った2進数に、新しい `0` または `1` を1桁追加する書き方**です。

この問題では、次のコードと同じ意味です。

```python
current = current * 2 + node.val
```

つまり、

```python
current = (current << 1) | node.val
```

は、2進数を扱うときによく使う高速な書き方です。

---

## `<<` の意味

`<<` は **左ビットシフト** です。

```python
current << 1
```

は、2進数を左に1桁ずらすという意味です。

10進数で考えると、

```python
current * 2
```

と同じです。

---

## 例：左ビットシフト

```python
current = 2
```

`2` は2進数で、

```text
10
```

です。

これを左に1桁ずらすと、

```text
10 << 1 = 100
```

になります。

`100` は10進数で `4` なので、

```python
2 << 1 == 4
```

です。

---

## `|` の意味

`|` は **ビットOR** です。

```python
a | b
```

は、2進数の各桁を比べて、どちらかが `1` なら `1` にします。

この問題では `node.val` は `0` または `1` なので、

```python
(current << 1) | node.val
```

は、

```text
左に1桁ずらしたあと、最後の桁に node.val を入れる
```

という意味になります。

---

## 例：`1 -> 0 -> 1` の場合

最初は、

```python
current = 0
```

です。

---

### 1つ目のノードが `1`

```python
current = (0 << 1) | 1
```

これは、

```text
0 に 1 を追加する
```

という意味です。

結果は、

```python
current = 1
```

です。

---

### 2つ目のノードが `0`

```python
current = (1 << 1) | 0
```

2進数で見ると、

```text
1 << 1 = 10
10 | 0 = 10
```

結果は、

```python
current = 2
```

です。

2進数では `10` です。

---

### 3つ目のノードが `1`

```python
current = (2 << 1) | 1
```

2進数で見ると、

```text
10 << 1 = 100
100 | 1 = 101
```

結果は、

```python
current = 5
```

です。

2進数では `101` です。

---

## 普通の書き方との比較

### わかりやすい書き方

```python
current = current * 2 + node.val
```

### ビット演算を使った書き方

```python
current = (current << 1) | node.val
```

この問題では、どちらも同じ意味です。

---

## なぜこの問題で使えるのか

根から葉までのパスが、

```text
1 -> 0 -> 1
```

なら、作りたい2進数は、

```text
101
```

です。

ノードを1つ進むたびに、2進数の桁を1つ追加したいので、

```python
current = current * 2 + node.val
```

または、

```python
current = (current << 1) | node.val
```

を使います。

---

## 覚え方

```python
current << 1
```

は、

```text
2進数を左に1桁ずらす
= 10進数では2倍する
```

です。

```python
| node.val
```

は、

```text
最後の桁に node.val を追加する
```

です。

したがって、

```python
current = (current << 1) | node.val
```

は、

```text
今までの2進数に、新しい0または1を1桁追加する
```

という意味です。

---

## この問題での実際のコード

```python
class Solution:
    def sumRootToLeaf(self, root: Optional[TreeNode]) -> int:
        def dfs(node, current):
            if not node:
                return 0

            current = (current << 1) | node.val

            if not node.left and not node.right:
                return current

            return dfs(node.left, current) + dfs(node.right, current)

        return dfs(root, 0)
```

---

## 計算量

```text
時間計算量: O(n)
空間計算量: O(h)
```

`n` はノード数、`h` は木の高さです。

全ノードを1回ずつ見るだけなので高速です。

# heap / heapq の学習事項まとめ

## heapとは

heap は、最小値や最大値を効率よく取り出すためのデータ構造。

Python の heapq は「最小ヒープ」なので、基本的には一番小さい値が先に取り出される。

つまり、

    heap[0]

には常に最小値が入る。

---

## heap は関数ではない

heap は関数ではなく、データ構造。

Python では heapq というモジュールを使って、普通のリストを heap として扱う。

    import heapq

---

## よく使う関数

| 関数 | 意味 |
|---|---|
| heapq.heappush(heap, value) | heap に値を追加する |
| heapq.heappop(heap) | heap から最小値を取り出す |
| heapq.heapify(list) | 普通のリストを heap に変換する |

---

## heappush

heap に値を追加する。

    import heapq

    heap = []

    heapq.heappush(heap, 5)
    heapq.heappush(heap, 2)
    heapq.heappush(heap, 8)

    print(heap)

heap の中身は完全な昇順になるとは限らない。

ただし、

    heap[0]

は必ず最小値になる。

---

## heappop

heap から最小値を取り出す。

    smallest = heapq.heappop(heap)

例えば heap に 2, 5, 8 が入っていたら、heappop で 2 が取り出される。

---

## heapify

普通のリストを heap に変換する。

    import heapq

    nums = [5, 2, 8, 1]

    heapq.heapify(nums)

    print(nums)

これで nums を heap として使えるようになる。

---

## 今回の問題で heap を使う理由

今回の問題では、毎回、

- 一番重い石
- 二番目に重い石

を取り出す必要がある。

毎回 sort すると時間がかかる。

heap を使うと、値の取り出しを効率よくできる。

---

## Python の heapq は最小ヒープ

Python の heapq は、一番小さい値を取り出す仕組み。

しかし今回ほしいのは、

    一番重い石 = 最大値

なので、そのままでは使いにくい。

そこで、石の重さをマイナスにして heap に入れる。

---

## 最大値を取り出す工夫

例えば、

    stones = [2, 7, 4, 1, 8, 1]

これをそのまま heap にすると、最小値の 1 が先に出る。

でも今回は最大値の 8 を先に取り出したい。

そこで、すべてマイナスにする。

    [-2, -7, -4, -1, -8, -1]

この中で一番小さい値は、

    -8

である。

これは元の重さに戻すと、

    8

になる。

つまり、マイナスにして入れることで、最小ヒープを最大ヒープのように使える。

---

## 今回のコード

    from typing import List
    import heapq

    class Solution:
        def lastStoneWeight(self, stones: List[int]) -> int:
            heap = []

            for stone in stones:
                heapq.heappush(heap, -stone)

            while len(heap) > 1:
                y = -heapq.heappop(heap)
                x = -heapq.heappop(heap)

                if x != y:
                    heapq.heappush(heap, -(y - x))

            if heap:
                return -heap[0]
            else:
                return 0

---

## コードの流れ

### 1. heap を作る

    heap = []

    for stone in stones:
        heapq.heappush(heap, -stone)

石の重さをマイナスにして heap に入れる。

これにより、重い石ほど先に取り出せる。

---

### 2. 石が2個以上ある間、処理を続ける

    while len(heap) > 1:

石が2個以上残っている間は、ゲームを続ける。

---

### 3. 一番重い石と二番目に重い石を取り出す

    y = -heapq.heappop(heap)
    x = -heapq.heappop(heap)

heapq.heappop(heap) は最小値を取り出す。

ただし、石はマイナスで入れている。

例えば、

    -8

を取り出したら、

    -(-8) = 8

として元の重さに戻す。

---

## x と y の意味

問題文では、

    x <= y

とされている。

つまり、

    y = 一番重い石
    x = 二番目に重い石

コードでは、

    y = -heapq.heappop(heap)
    x = -heapq.heappop(heap)

となっている。

最初に取り出されるのが一番重い石なので y。

次に取り出されるのが二番目に重い石なので x。

---

## 4. 石の重さが違う場合

    if x != y:
        heapq.heappush(heap, -(y - x))

例えば、

    y = 8
    x = 7

なら、

    y - x = 1

なので、重さ 1 の石が残る。

ただし heap にはマイナスで入れるので、

    heapq.heappush(heap, -1)

とする。

## 5. 最後の石を返す

    if heap:
        return -heap[0]
    else:
        return 0

最後に石が残っていれば、その重さを返す。

ただし heap にはマイナスで入っているので、

    -heap[0]

として正の値に戻す。

石が残っていなければ 0 を返す。

---

## 重要ポイント

- heap は最小値を効率よく取り出せるデータ構造
- Python の heapq は最小ヒープ
- 最大値を取り出したいときは、値をマイナスにして入れる
- heappop で取り出したあと、もう一度マイナスをつけて元に戻す
- 毎回最大値を2つ取り出す問題では heap が便利

## 計算量

石の数を n とする。

時間計算量:

    O(n log n)

空間計算量:

    O(n)

heap への追加や取り出しには O(log n) かかる。

最大で n 個の石を処理するので、全体では O(n log n) になる。

今回の問題は、

    毎回、一番大きい値を2つ取り出す

そのため、

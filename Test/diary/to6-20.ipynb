# 学習事項：`bit_count()` について

## 1. `bit_count()` とは

Pythonには、整数を2進数で見たときの `1` の個数を数えるメソッドがある。

    x.bit_count()

これは、整数 `x` の2進数表現に含まれる `1` の数を返す。

---

## 2. 例

    x = 8
    print(x.bit_count())

`8` は2進数で表すと、

    1000

なので、`1` は1個。

結果：

    1

---

別の例：

    x = 7
    print(x.bit_count())

`7` は2進数で表すと、

    111

なので、`1` は3個。

結果：

    3

---

## 3. 数字に直接使う場合の注意

変数に入っている整数なら、普通にこう書ける。

    x = 8
    x.bit_count()

ただし、数字を直接書く場合は、

    8.bit_count()

ではなく、

    (8).bit_count()

のように書く方が安全。

理由は、Pythonでは数字の直後の `.` が小数点と紛らわしいから。

---

## 4. 今回の問題での使い方

問題：

整数を、2進数にしたときの `1` の個数が少ない順に並べる。

同じ個数なら、数字が小さい順に並べる。

コード：

    class Solution:
        def sortByBits(self, arr: List[int]) -> List[int]:
            return sorted(arr, key=lambda x: (x.bit_count(), x))

---

## 5. `(x.bit_count(), x)` の意味

    key=lambda x: (x.bit_count(), x)

これは、ソートの基準を表している。

タプルは、左から順番に比較される。

つまり、

    (x.bit_count(), x)

は、

    第1基準：2進数にしたときの 1 の個数
    第2基準：数字そのものの大きさ

という意味。

---

## 6. 具体例

    arr = [0, 1, 2, 3, 4, 5, 6, 7, 8]

それぞれの2進数と `1` の個数は以下。

| 数字 | 2進数 | 1の個数 |
|---|---|---|
| 0 | 0 | 0 |
| 1 | 1 | 1 |
| 2 | 10 | 1 |
| 3 | 11 | 2 |
| 4 | 100 | 1 |
| 5 | 101 | 2 |
| 6 | 110 | 2 |
| 7 | 111 | 3 |
| 8 | 1000 | 1 |

したがって、並べ替えると、

    [0, 1, 2, 4, 8, 3, 5, 6, 7]

になる。

---

## 7. 古い書き方

`bit_count()` を使わない場合は、次のようにも書ける。

    bin(x).count("1")

例：

    x = 7
    bin(x)             # "0b111"
    bin(x).count("1")  # 3

ただし、今回のような問題では、

    x.bit_count()

の方が短くて速い。

---

## 8. まとめ

- `x.bit_count()` は、整数 `x` の2進数表現に含まれる `1` の個数を返す。
- 例：`8 = 1000` なので `(8).bit_count()` は `1`。
- 例：`7 = 111` なので `(7).bit_count()` は `3`。
- 数字に直接使う場合は `(8).bit_count()` のようにカッコをつける。
- `sorted(arr, key=lambda x: (x.bit_count(), x))` と書くと、
  - まず `1` の個数が少ない順
  - 同じなら数字が小さい順
  に並べられる。

# 学習事項：`datetime.strptime()` はPythonの組み込み関数？

## 結論

`datetime.strptime()` は **Pythonの組み込み関数ではない**。

ただし、Pythonに最初から入っている **標準ライブラリ** の機能。

つまり、追加インストールは不要だが、使うには `import` が必要。

---

## 組み込み関数とは？

組み込み関数とは、`import` しなくてもそのまま使える関数のこと。

例：

- `abs()`
- `len()`
- `sum()`
- `max()`
- `min()`
- `range()`
- `print()`

例えば今回のコードの中では、

    abs((d1 - d2).days)

の `abs()` は組み込み関数。

`abs()` は絶対値を返す関数。

例：

    abs(-15)
    # 15

---

## `datetime` とは？

`datetime` はPythonの標準ライブラリ。

日付や時刻を扱うための機能が入っている。

使うには次のように `import` する。

    from datetime import datetime

---

## `strptime()` とは？

`strptime()` は、文字列を日付データに変換する関数。

今回なら、

    d1 = datetime.strptime(date1, "%Y-%m-%d")

によって、

    "2019-06-29"

という文字列を、日付として扱えるデータに変換している。

---

## `%Y-%m-%d` の意味

    "%Y-%m-%d"

は、日付文字列の形式を指定している。

| 記号 | 意味 | 例 |
|---|---|---|
| `%Y` | 4桁の年 | `2019` |
| `%m` | 2桁の月 | `06` |
| `%d` | 2桁の日 | `29` |

つまり、

    "%Y-%m-%d"

は、

    2019-06-29

のような形式を表している。

---

## 今回のコード

    from datetime import datetime

    class Solution:
        def daysBetweenDates(self, date1: str, date2: str) -> int:
            d1 = datetime.strptime(date1, "%Y-%m-%d")
            d2 = datetime.strptime(date2, "%Y-%m-%d")

            return abs((d1 - d2).days)

---

## コードの流れ

例えば、

    date1 = "2020-01-15"
    date2 = "2019-12-31"

のとき、

まず文字列を日付データに変換する。

    d1 = datetime.strptime(date1, "%Y-%m-%d")
    d2 = datetime.strptime(date2, "%Y-%m-%d")

そのあと、日付同士を引き算する。

    d1 - d2

すると、日数差を表すデータが得られる。

その中の `.days` を使うと、日数だけを取り出せる。

    (d1 - d2).days

ただし、日付の順番によってはマイナスになる。

そのため、`abs()` を使って絶対値にする。

    abs((d1 - d2).days)

---

## 重要ポイント

- `datetime.strptime()` は組み込み関数ではない
- `datetime` は標準ライブラリなので追加インストール不要
- ただし `from datetime import datetime` が必要
- `strptime()` は文字列を日付に変換する
- `abs()` は組み込み関数
- `.days` で日付差の日数部分を取り出せる

---

## 覚え方

文字列の日付を計算したいときは、

    datetime.strptime(日付文字列, 日付の形式)

を使う。

日数差を出すときは、

    abs((日付1 - 日付2).days)

を使う。

# SQLのエイリアスまとめ

## エイリアスとは

SQLのエイリアスとは、テーブル名やカラム名に一時的な別名をつけること。

特にテーブル名が長いときに、短い名前で書けるようになる。

例：

    FROM Employees e

これは、

    Employees テーブルを e という名前で扱う

という意味。

---

## 今回のSQL

    SELECT 
        u.unique_id,
        e.name
    FROM Employees e
    LEFT JOIN EmployeeUNI u
    ON e.id = u.id;

ここで使われているエイリアスは次の2つ。

| エイリアス | 元のテーブル名 |
|---|---|
| e | Employees |
| u | EmployeeUNI |

つまり、

    e.id

は、

    Employees.id

のこと。

また、

    u.id

は、

    EmployeeUNI.id

のこと。

---

## エイリアスを使わずに書いた場合

    SELECT 
        EmployeeUNI.unique_id,
        Employees.name
    FROM Employees
    LEFT JOIN EmployeeUNI
    ON Employees.id = EmployeeUNI.id;

これでも正解。

ただし、テーブル名を毎回全部書くので長くなる。

---

## エイリアスを使った場合

    SELECT 
        u.unique_id,
        e.name
    FROM Employees e
    LEFT JOIN EmployeeUNI u
    ON e.id = u.id;

短く書けて読みやすい。

---

## ON e.id = u.id の意味

    ON e.id = u.id

これは、

    Employees テーブルの id と
    EmployeeUNI テーブルの id が同じ行を結合する

という意味。

エイリアスを元に戻すと、

    ON Employees.id = EmployeeUNI.id

になる。

---

## なぜエイリアスを使うのか

主な理由は3つ。

1. SQLを短く書ける
2. どのテーブルのカラムか分かりやすくなる
3. 同じ名前のカラムが複数テーブルにあるときに区別できる

今回だと、両方のテーブルに id カラムがある。

Employees にも id があり、EmployeeUNI にも id がある。

そのため、ただ

    id

と書くと、どちらの id なのか分かりにくい。

そこで、

    e.id
    u.id

のように書くことで、どのテーブルの id かを明確にしている。

---

## カラム名にもエイリアスをつけられる

テーブルだけでなく、カラムにもエイリアスをつけられる。

例：

    SELECT 
        name AS employee_name
    FROM Employees;

これは、

    name カラムを employee_name という名前で表示する

という意味。

出力結果の列名が employee_name になる。

---

## AS は省略できる

テーブルのエイリアスでは、AS を省略することが多い。

次の2つはほぼ同じ意味。

    FROM Employees AS e

    FROM Employees e

LeetCodeなどでは、短く

    FROM Employees e

と書くことが多い。

---

## 今回の重要ポイント

    Employees e

は、

    Employees を e と呼ぶ

という意味。

    EmployeeUNI u

は、

    EmployeeUNI を u と呼ぶ

という意味。

したがって、

    e.name

は Employees の name。

    u.unique_id

は EmployeeUNI の unique_id。

    e.id = u.id

は Employees と EmployeeUNI を id で結合する条件。

---

## 覚え方

    テーブル名 エイリアス名

の形で覚える。

例：

    Employees e

意味：

    Employees を e として使う

例：

    EmployeeUNI u

意味：

    EmployeeUNI を u として使う

エイリアスは、SQLを短く、分かりやすく書くための省略名。

# `zip(*matrix)` について

## 1. `zip()` とは

`zip()` は、複数のリストを「同じ位置どうし」でまとめる関数です。

例えば：

    a = [1, 2, 3]
    b = [4, 5, 6]

    print(list(zip(a, b)))

結果：

    [(1, 4), (2, 5), (3, 6)]

つまり、

| a | b | まとめた結果 |
|---|---|---|
| 1 | 4 | (1, 4) |
| 2 | 5 | (2, 5) |
| 3 | 6 | (3, 6) |

のように、同じインデックス同士をペアにします。

---

## 2. `*matrix` とは

`*` は、リストの中身をバラして渡すための記法です。

例えば：

    matrix = [
        [3, 7, 8],
        [9, 11, 13],
        [15, 16, 17]
    ]

この `matrix` は、実際には次のようなリストです。

    [
        [3, 7, 8],
        [9, 11, 13],
        [15, 16, 17]
    ]

ここで、

    zip(*matrix)

と書くと、これは次のように書いたのと同じ意味になります。

    zip(
        [3, 7, 8],
        [9, 11, 13],
        [15, 16, 17]
    )

つまり、`matrix` の各行をバラして `zip()` に渡しています。

---

## 3. `zip(*matrix)` の動き

例：

    matrix = [
        [3, 7, 8],
        [9, 11, 13],
        [15, 16, 17]
    ]

このとき、

    list(zip(*matrix))

の結果は：

    [
        (3, 9, 15),
        (7, 11, 16),
        (8, 13, 17)
    ]

になります。

これは、行列を「列ごと」に取り出しているということです。

元の行列：

|      | 列0 | 列1 | 列2 |
|------|-----|-----|-----|
| 行0 | 3   | 7   | 8   |
| 行1 | 9   | 11  | 13  |
| 行2 | 15  | 16  | 17  |

`zip(*matrix)` の結果：

| 列 | 取り出される値 |
|---|---|
| 列0 | (3, 9, 15) |
| 列1 | (7, 11, 16) |
| 列2 | (8, 13, 17) |

つまり、`zip(*matrix)` は行列の「列」を取り出すときによく使います。

---

## 4. なぜラッキーナンバー問題で使うのか

ラッキーナンバーは、

- その行では最小値
- その列では最大値

である数です。

行の最小値は簡単に求められます。

    row_mins = set(min(row) for row in matrix)

例えば：

    matrix = [
        [3, 7, 8],
        [9, 11, 13],
        [15, 16, 17]
    ]

各行の最小値は：

    3, 9, 15

なので、

    row_mins = {3, 9, 15}

になります。

一方、列の最大値を求めるには、列ごとに見たいです。

そのために、

    zip(*matrix)

を使います。

    zip(*matrix)

は、

    (3, 9, 15)
    (7, 11, 16)
    (8, 13, 17)

を作ってくれるので、それぞれに `max()` を使えば列の最大値が求まります。

    col_maxs = set(max(col) for col in zip(*matrix))

結果：

    col_maxs = {15, 16, 17}

---

## 5. コード全体

    from typing import List

    class Solution:
        def luckyNumbers(self, matrix: List[List[int]]) -> List[int]:
            # 各行の最小値を集める
            row_mins = set(min(row) for row in matrix)

            # zip(*matrix) で列ごとに取り出し、各列の最大値を集める
            col_maxs = set(max(col) for col in zip(*matrix))

            # 行の最小値 かつ 列の最大値になっている数を返す
            return list(row_mins & col_maxs)

---

## 6. `row_mins & col_maxs` の意味

`&` は、集合の共通部分を求める演算子です。

例えば：

    row_mins = {3, 9, 15}
    col_maxs = {15, 16, 17}

このとき、

    row_mins & col_maxs

は、

    {15}

になります。

つまり、

    行の最小値でもあり、列の最大値でもある数

を取り出しています。

---

## 7. まとめ

`zip(*matrix)` は、二次元リストを列ごとに取り出すための書き方です。

重要なイメージ：

    matrix = [
        [3, 7, 8],
        [9, 11, 13],
        [15, 16, 17]
    ]

    zip(*matrix)

結果：

    (3, 9, 15)
    (7, 11, 16)
    (8, 13, 17)

つまり、

    行方向のデータを、列方向に変換している

と考えるとわかりやすいです。

ラッキーナンバー問題では、各列の最大値を求めたいので `zip(*matrix)` を使います。

# `ok = True` の意味

## 1. `ok = True` とは

`ok = True` は、

    「今のところ条件を満たしている」

という意味の目印です。

このような変数を **フラグ変数** と呼びます。

    ok = True

と書くことで、最初は

    問題なし
    条件を満たしている
    カウントしてよさそう

という状態にしておきます。

---

## 2. 今回の問題での意味

今回の問題では、`arr1` の各要素 `x` について、

    arr2 の中に
    |x - y| <= d
    を満たす y が1つもない

ならカウントします。

つまり、`x` が `arr2` のすべての要素と十分に離れていればOKです。

そのため、最初に

    ok = True

としておきます。

これは、

    この x は今のところカウントしてよい

という意味です。

---

## 3. コードの流れ

    count = 0

    for x in arr1:
        ok = True

        for y in arr2:
            if abs(x - y) <= d:
                ok = False
                break

        if ok:
            count += 1

    return count

---

## 4. `ok = True` から始める理由

`arr2` の中を調べる前は、まだ `x` がダメかどうかわかりません。

なので最初は仮に、

    ok = True

として、

    この x は条件を満たしているはず

という前提で始めます。

その後、`arr2` の中に近すぎる値が見つかったら、

    ok = False

に変更します。

---

## 5. `ok = False` になる条件

今回の条件はこれです。

    if abs(x - y) <= d:

これは、

    x と y の距離が d 以下

という意味です。

つまり、

    x に近すぎる y が arr2 に存在する

ということです。

その場合、その `x` はカウントできません。

だから、

    ok = False

にします。

---

## 6. `break` の意味

    if abs(x - y) <= d:
        ok = False
        break

近すぎる値が1つでも見つかった時点で、その `x` はもうカウントできません。

そのため、それ以上 `arr2` を調べる必要がありません。

なので、

    break

で内側の `for y in arr2` を途中で終了します。

---

## 7. `if ok:` の意味

    if ok:
        count += 1

これは、

    ok が True のままだったら count を1増やす

という意味です。

つまり、

    arr2 の中に近すぎる値が1つもなかった

ということです。

だから、その `x` は distance value にカウントされます。

---

## 8. 具体例

    arr1 = [4, 5, 8]
    arr2 = [10, 9, 1, 8]
    d = 2

### x = 4 の場合

最初：

    ok = True

調べる：

    |4 - 10| = 6 > 2
    |4 - 9|  = 5 > 2
    |4 - 1|  = 3 > 2
    |4 - 8|  = 4 > 2

どれも `<= 2` ではありません。

つまり、近すぎる値がないので、

    ok = True

のままです。

よって、

    count += 1

します。

---

### x = 8 の場合

最初：

    ok = True

調べる：

    |8 - 10| = 2 <= 2

ここで条件に引っかかります。

なので、

    ok = False
    break

になります。

この時点で `x = 8` はカウントできません。

---

## 9. `all()` を使った書き方との関係

このコード：

    ok = True

    for y in arr2:
        if abs(x - y) <= d:
            ok = False
            break

は、次のコードと同じ意味です。

    all(abs(x - y) > d for y in arr2)

つまり、

    ok = True を使う書き方
    all() を使う書き方

は、どちらも

    arr2 のすべての要素 y について
    abs(x - y) > d が成り立つか

を判定しています。

---

## 10. まとめ

`ok = True` は、条件判定のための目印です。

今回の問題では、

    ok = True
    → この x は今のところカウントしてよい

という意味です。

そして、近すぎる値が見つかったら、

    ok = False

にして、

    この x はカウントしない

という状態に変更します。

最後に、

    if ok:
        count += 1

とすることで、

    最後まで ok が True のままだった x だけを数える

ことができます。

重要なイメージ：

    ok = True から始める
    ダメな条件が見つかったら ok = False にする
    最後まで True ならカウントする

# 学習事項：`startswith()` について

## 何をするものか

`startswith()` は、Pythonにもともと用意されている **文字列用のメソッド**。

自分で定義する関数ではなく、文字列に対して使える標準機能。

```python
word.startswith(searchWord)
```

これは、

```text
word が searchWord で始まっているか
```

を判定する。

戻り値は `True` または `False`。

---

## 例

```python
"burger".startswith("burg")
```

`"burger"` は `"burg"` で始まっているので、

```python
True
```

になる。

一方で、

```python
"burger".startswith("ger")
```

`"ger"` は `"burger"` の中にはあるが、先頭ではないので、

```python
False
```

になる。

---

## 今回の問題での使い方

```python
if word.startswith(searchWord):
```

これは、

```text
もし word が searchWord で始まっていたら
```

という意味。

例えば、

```python
word = "burger"
searchWord = "burg"
```

なら、

```python
word.startswith(searchWord)
```

は `True` になる。

---

## `startswith()` を使わずに書くと

`startswith()` は、次のような処理とほぼ同じ意味。

```python
word[:len(searchWord)] == searchWord
```

例えば、

```python
word = "burger"
searchWord = "burg"
```

の場合、

```python
len(searchWord)
```

は `4`。

なので、

```python
word[:len(searchWord)]
```

は、

```python
word[:4]
```

と同じで、結果は

```python
"burg"
```

になる。

つまり、

```python
word[:len(searchWord)] == searchWord
```

は、

```python
"burg" == "burg"
```

となり、`True` になる。

---

## まとめ

`startswith()` は、文字列が指定した文字列で始まっているかを調べるメソッド。

```python
word.startswith(searchWord)
```

は、

```python
word[:len(searchWord)] == searchWord
```

とほぼ同じ意味。

今回の問題では、各単語が `searchWord` で始まっているかを確認するために使う。
